```python
import xarray as xr
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os
```

```python
file_path = r"./.data/v04r01/access/netcdf/IBTrACS.WP.v04r01.nc"
# Check that the file exists
print("File exists:", os.path.exists(file_path))

ds = xr.open_dataset(file_path, engine="netcdf4")
ds
```

```python
required_vars = [
    "sid",
    "name",
    "season",
    "time",
    "cma_lat",
    "cma_lon",
    "cma_wind",
    "cma_pres",
    "cma_cat",
    "source_cma",
    "iflag",
    "track_type",
    "dist2land",
    "landfall"
]

# Select variables that actually exist in your dataset
available_vars = [
    v for v in required_vars
    if v in ds.variables
]

ds = ds[available_vars]

ds
```

TCs from 1979 to 2022


```python
# Select tropical cyclones from 1979 to 2022
year_mask = (
    (ds["season"] >= 1979) &
    (ds["season"] <= 2022)
)

ds_filtered = ds.where(year_mask, drop=True)
ds_filtered
```

```python
# First storm, first observation
record = ds_filtered.isel(storm=0, date_time=0)
record
```
